```python
X_A = df.drop(columns=['unit','cycle','RUL'])
y = df['RUL']
print('Feature shape',X_A.shape)
print('Target shape',y.shape)
```

```python
from sklearn.model_selection import GroupShuffleSplit 
splitter = GroupShuffleSplit(n_splits=1,test_size=0.2,random_state=42)
train_idx ,val_idx = next(splitter.split(X_A,y,groups=df['unit']))
```

```python
X_train_A = X_A.iloc[train_idx]
X_val_A = X_A.iloc[val_idx]
y_train = y.iloc[train_idx]
y_val = y.iloc[val_idx]
```

```python
print("Training engines:", df.iloc[train_idx]['unit'].nunique())
print("Validation engines:", df.iloc[val_idx]['unit'].nunique())
print("Training rows:", len(X_train_A))
print("validation rows:", len(X_val_A))
```

```python
from sklearn.ensemble import RandomForestRegressor
model_A = RandomForestRegressor(n_estimators=100,random_state=42,n_jobs=-1)
model_A.fit(X_train_A,y_train)
```

```python
from sklearn.metrics import mean_absolute_error,root_mean_squared_error
y_pred_A = model_A.predict(X_val_A)

MAE=mean_absolute_error(y_val,y_pred_A)
RMSE=root_mean_squared_error(y_val,y_pred_A)

print(f"MAE : {MAE}\nRMSE : {RMSE}")
```

```python
plt.figure(figsize=(8,5))
plt.scatter(y_val,y_pred_A,alpha=0.5)
plt.xlabel("Actual RUL")
plt.ylabel("Predicted RUL")
plt.title("Actual vs Predicted RUL - Approach A")
plt.show()
```

```python
X_B = df.drop(columns=['unit','RUL'])
y = df['RUL']
print("Feature shape:",X_B.shape)
```

```python
from sklearn.model_selection import GroupShuffleSplit 
splitter = GroupShuffleSplit(n_splits=1,test_size=0.2,random_state=42)
train_idx ,val_idx = next(splitter.split(X_B,y,groups=df['unit']))
```

```python
X_train_B = X_B.iloc[train_idx]
X_val_B = X_B.iloc[val_idx]
y_train = y.iloc[train_idx]
y_val = y.iloc[val_idx]
```

```python
print("Training engines:", df.iloc[train_idx]['unit'].nunique())
print("Validation engines:", df.iloc[val_idx]['unit'].nunique())
print("Training rows:", len(X_train_B))
print("validation rows:", len(X_val_B))
```

```python
from sklearn.ensemble import RandomForestRegressor
model_B = RandomForestRegressor(n_estimators=300,random_state=42,n_jobs=-1)
model_B.fit(X_train_B,y_train)
```

```python
from sklearn.metrics import mean_absolute_error,root_mean_squared_error
y_pred_B = model_B.predict(X_val_B)

MAE=mean_absolute_error(y_val,y_pred_B)
RMSE=root_mean_squared_error(y_val,y_pred_B)

print(f"MAE : {MAE}\nRMSE : {RMSE}")
```

```python
line = [x for x in range(300)]
plt.figure(figsize=(8,5))
plt.scatter(y_val,y_pred_B,alpha=0.5,s=6)
plt.plot(line,line,color='orange')
plt.xlabel("Actual RUL")
plt.ylabel("Predicted RUL")
plt.title("Actual vs Predicted RUL - Approach B")
plt.show()
```

```python
limit = max(y_val.max(), y_pred_B.max())
plt.figure(figsize=(8,5))
plt.scatter(y_val,y_pred_B,alpha=0.4,s=0.7) # alpha: for transperncy degree,s: for point size
plt.plot([0, limit], [0, limit], color='orange', label='Perfect Prediction')
plt.xlabel("Actual RUL")
plt.ylabel("Predicted RUL")
plt.title("Actual vs Predicted RUL - Approach B")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

importance = pd.Series(
    model_B.feature_importances_,
    index=X_B.columns
).sort_values(ascending=False)

print(importance)
```

```python
importance.sort_values().plot(kind='barh', figsize=(8, 6))
plt.title("Feature Importance — Approach B")
plt.xlabel("Importance")
plt.show()
```

```python
# Select the top 10 important features
top_features = importance.head(10).index.tolist()

print("Selected features:", top_features)
```

```python
X_train_C = X_train_B[top_features]
X_val_C = X_val_B[top_features]

print("Training shape:", X_train_C.shape)
print("Validation shape:", X_val_C.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor
model_C = RandomForestRegressor(n_estimators=100,random_state=42,n_jobs=-1)
model_C.fit(X_train_C,y_train)
from sklearn.metrics import mean_absolute_error,root_mean_squared_error
y_pred_C = model_C.predict(X_val_C)

MAE=mean_absolute_error(y_val,y_pred_C)
RMSE=root_mean_squared_error(y_val,y_pred_C)

print(f"MAE : {MAE}\nRMSE : {RMSE}")
```

```python
line = [x for x in range(300)]
plt.figure(figsize=(8,5))
plt.scatter(y_val,y_pred_C,alpha=0.5,s=6)
plt.plot(line,line,color='orange')
plt.xlabel("Actual RUL")
plt.ylabel("Predicted RUL")
plt.title("Actual vs Predicted RUL - Approach C")
plt.show()
```
